### 1. Users

```python
all_user_df = pl.read_csv('datalake/clean_merge/users.csv')
all_user_df
```

```python
total = len(all_user_df)
for i, row in enumerate(all_user_df.rows(named=True)):
    try:
        record = client.collection('user_account').create(row);
    except Exception as e:
        print(f'Error: {e.data}')
    print(f"\r{i}/{total} ", end='', flush=True)
```
