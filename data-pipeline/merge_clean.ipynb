```python
import polars as pl
import pandas as pd
```

```python
goodreads_book_df = pl.read_csv('datalake/clean/cleaned_books_data.csv').rename({
    'isbn13':'id', 'isbn':'ISBN_10','crawl_source':'crawl_source_url'
})

# Clean the string
goodreads_book_df = goodreads_book_df.with_columns(
    # Split errored ids
    pl.col('id')
    .str.split(',')
    .list.eval(pl.element().str.strip_chars())
).explode('id').with_columns(
    [
        pl.col(col).str.strip_chars("'").str.strip_chars('"').alias(col)
        for col in goodreads_book_df.columns if goodreads_book_df.schema[col] == pl.Utf8
    ]
)

# Add source column
goodreads_book_df = goodreads_book_df.with_columns(
    pl.lit('goodreads').alias('crawl_source')
)

goodreads_book_df
```

```python
# Further cleaning
# Change publish date to string date
goodreads_book_df = goodreads_book_df.with_columns(
    pl.col("publish_date")
    .cast(pl.Datetime("ms")) # Convert to Datetime
    .dt.strftime("%Y-%m-%d") # Format as "yyyy-mm-dd"
)

goodreads_book_df = goodreads_book_df.filter(pl.col("id") != "None")
```

```python
hardcover_genre_df = pl.read_csv('datalake/hardcover/hardcover_book_genre.csv', schema={
    "id": pl.String,
    "genre": pl.String,
}).rename({'id':'isbn13'})
hardcover_genre_df
```

Lots of non-text characters such as ", #, &, ', (), ., etc

```python
unusual_df = hardcover_genre_df.filter(pl.col("genre").str.contains(r"[^a-zA-Z0-9 ]")).unique(subset="genre")
unusual_df.write_csv('./genre_analysis.csv')
unusual_df
```

```python
def split_outside_parentheses(text):
    result = []
    current = []
    paren_count = 0

    for char in text:
        if char == '(':
            paren_count += 1
        elif char == ')':
            paren_count -= 1
        elif char == ',' and paren_count == 0:
            if len(current) == 0:
                continue
            # Split here if outside parentheses
            result.append(''.join(current).strip())
            current = []
            continue
        current.append(char)

    # if open and not closed
    if paren_count != 0:
        return []
        
    # Append the last part
    result.append(''.join(current).strip())
    
    return result
```

```python
special_cases = {
    'Body, Mind & Spirit': 'Body Mind and Spirit',
    'Comic books, strips, etc': ', Comics',
    'comic books': ', Comic books',
    'Babel, Tower of': 'Tower of Babel',
    'Womens ': "Women's ",

    'in literature': '',
    'Adventure and adventures': 'Adventure and adventurers',
    'Archaeology and': 'Archaeology,',
    "Boy's Love": 'Boys Love',
    'Detective and mystery': 'Detective and mystery,'
# TODO: Remove plurals, e.g. Boarding schools -> Boarding school, Books and readings -> Books and reading, but not "materials"
}

hardcover_genre_df_fixed = hardcover_genre_df.with_columns(
    pl.col('genre')
    .str.replace_all(r'[.\"ʹ]','')             # Remove all . "
    .str.replace_all('’',"'",literal=True)    # Replace weird quotation ’
    .str.replace_many(pl.Series(special_cases.keys()), pl.Series(special_cases.values()), ascii_case_insensitive=True) # Replace special cases
    .str.replace_all(r'\/|--| - |\|','|')     # Replace some OR symbols like / -- | " - " and split them
    .str.split('|')
    .list.eval(pl.element().str.strip_chars())
    .alias('split_genres')
).explode('split_genres').drop(
    'genre'
).with_columns(
    pl.col("split_genres")
    .map_elements(split_outside_parentheses, return_dtype=pl.List(pl.String)).alias("genre")
).explode('genre').drop_nulls().drop('split_genres')

hardcover_genre_df_fixed
```

```python
unique_chars = set(''.join(hardcover_genre_df_fixed['genre']))
print("Unique Characters:")
print(sorted(unique_chars))
```

```python
goodreads_genre_df = pl.read_csv('datalake/clean/cleaned_books_genres_data.csv').rename({"genre_name":"genre"})

# Strip leading quotes
goodreads_genre_df = goodreads_genre_df.with_columns(
    # Split isbn13 that isn't normal
    pl.col('isbn13')
    .str.split(',')
    .list.eval(pl.element().str.strip_chars())
).explode('isbn13').with_columns(
    # Strip quotes '
    [
        pl.col(col).str.strip_chars("'").str.strip_chars('"').alias(col)
        for col in goodreads_genre_df.columns if goodreads_genre_df.schema[col] == pl.Utf8
    ]
# Remove 'None' and stuffs
).filter(pl.col("isbn13").str.len_chars() == 13)

goodreads_genre_df
```

```python
unique_chars = set(''.join(goodreads_genre_df['genre']))
print("Unique Characters:")
print(sorted(unique_chars))
```

No non-text characters

```python
merged_genres = pl.concat([
    goodreads_genre_df.select(["isbn13","genre"]),
    hardcover_genre_df_fixed.select(["isbn13","genre"])
], how="vertical_relaxed")

print("Before: ", merged_genres.shape)

genre_wrong_cases = merged_genres.select(["genre"]).unique().with_columns(
    pl.col('genre')
    .str.to_lowercase()
    .str.replace_all(r'[&\'"()+\-]', "")
    .is_duplicated().alias('duplicated')
)

# Check duplicate if lowercased and convert accordingly
merged_genres_stripped = merged_genres.join(genre_wrong_cases, on='genre').with_columns(
    pl.when(pl.col('duplicated') == True)
    .then(pl.col('genre').str.to_titlecase())
    .otherwise(pl.col('genre'))
    .alias('genre')
).drop('duplicated').unique().filter(
    ~pl.col('isbn13').str.contains(r"[^0-9xX]") # Remove weird isbn13              
).filter(
    pl.col('genre').str.len_chars() > 0         # Non-empty
)

print(merged_genres_stripped)
merged_genres_stripped.describe()
```

```python
merged_genres_stripped.write_csv('datalake/clean_merge/book_genres.csv')
```

## Convert to valid SQLite csv format for Pocketbase

```python
from datetime import datetime, timezone
today_iso = datetime.now(timezone.utc).strftime("%Y-%m-%d %H:%M:%S.%f")[:-3] + "Z"

print(today_iso)
```

### 1. Genres

```python
genre_df = pl.read_csv('datalake/clean_merge/book_genres.csv', schema={
    'isbn13': pl.String,
    'genre_name': pl.String
}).rename({
    'genre_name':'name'
})

genre_df
```

##### Making ids

```python
from unidecode import unidecode
```

```python
genre_stripped_df = genre_df.group_by("name").agg(
    pl.len().alias("count")
).sort(
    "count", descending=True
).with_row_index(
    "id", offset=1
).with_columns(
    pl.lit(today_iso).alias("created")
)

genre_stripped_df
```

```python
genre_stripped_df.select(['created', 'id', 'name']).write_csv('../database/genre_toimport.csv')
```

```python
genre_df_with_id = genre_df.join(genre_stripped_df.select(['id','name']), on='name', how='left')
genre_df_with_id
```

```python
genre_df_joined = genre_df_with_id.group_by('isbn13').agg(
    pl.format("[{}]", pl.format('"{}"', pl.col('id')).str.join(',')).alias('joined_ids')
).rename({
 'joined_ids': 'genres',
 'isbn13':'id', 
})

genre_df_joined
```
