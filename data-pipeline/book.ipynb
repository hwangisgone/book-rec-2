<h2>Lấy genres của sách </h2>

```python

genres = []
# Khởi tạo trình duyệt
driver = webdriver.Chrome()

# Mở trang web
url = 'https://www.goodreads.com/book/show/7366'  # Thay bằng URL thực tế của bạn
driver.get(url)

try:
    # Tìm nút "more" bằng cách sử dụng CSS Selector chính xác
    more_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.XPATH, '//button[@aria-label="Show all items in the list"]')))
   
    more_button.click()
    
    # Sau khi nhấp, thực hiện các thao tác tiếp theo như lấy dữ liệu
    # Ví dụ: lấy nội dung trang sau khi đã nhấn nút "more"
    # page_source = driver.page_source
    # print(page_source)
    # print("Đã nhấn nút 'more' thành công.")
    
    book_list = WebDriverWait(driver, 10).until(
        EC.presence_of_all_elements_located((By.CLASS_NAME, "BookPageMetadataSection__genreButton"))
    )
    
    # Lấy thông tin từ phần tử có class "book_list"
    for index, element in enumerate(book_list, start=1):
        print(f"Thông tin book_list {index}:")
        print(element.text)
        genres.append(element.text)
        print("--------------------")
        
    # print(genres)
    
finally:
    # Đóng trình duyệt
    driver.quit()
```
