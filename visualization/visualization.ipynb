##### 2. Book distribution according to author

```python
author_count = defaultdict(int)
for author in books['author']:
    author_count[author] += 1

# Sắp xếp tác giả theo count giảm dần
author_count = sorted(author_count.items(), key=lambda x: x[1], reverse=True)

n = 150  # Độ dài của danh sách
zeros_list = [0] * n

# In kết quả
# for author, count in author_count:
#     # print(f"{author}: {count}")
#     zeros_list[count] += 1
        
# print(zeros_list)
```

```python
cnt_au_1 = 0
cnt_au_2 = 0
cnt_au_3 = 0
cnt_au_4 = 0

a1 = 5
a2 = 12
a3 = 24

for author, count in author_count:
    if count < a1:
        cnt_au_1 += 1
    elif a1 <= count <= a2:
        cnt_au_2  += 1
    elif a2 < count <= a3:
        cnt_au_3  += 1
    else:
        cnt_au_4  += 1
        
labels = [f'1 - {a1-1}', f'{a1} - {a2}', f'{a2+1} - {a3}', f'> {a3}']
sizes = [cnt_au_1 , cnt_au_2 , cnt_au_3 ,cnt_au_4]
        
plt.bar(labels, sizes, color='skyblue')

# Thêm nhãn
plt.xlabel("Number of books written by each author")
plt.ylabel('Count')
plt.title('Distribution of author based on the number of books one has written')

for i, size in enumerate(sizes):
    plt.text(i, size + 0.1, str(size), ha='center', va='bottom', fontsize=10)


# Hiển thị biểu đồ
plt.show()

# Danh sách các nhóm và số lượng tương ứng


# fig, ax = plt.subplots()
# wedges, texts, autotexts = ax.pie(
#     sizes,
#     autopct='%1.1f%%',
#     startangle=90,
#     pctdistance=0.5  # Đặt khoảng cách phần trăm bên trong biểu đồ
# )

# # Đặt tỷ lệ cân đối
# plt.axis('equal')

# # Chuyển chú thích sang bên phải
# ax.legend(
#     wedges,
#     [f'{label}: {size} authors' for label, size in zip(labels, sizes)],
#     loc="center left",
#     bbox_to_anchor=(1, 0, 0.5, 1)
# )

# # Thêm tiêu đề
# plt.title('Distribution of Authors Based on Book Count')

# # Hiển thị biểu đồ
# plt.show()
```

##### 3. Genre distribution

```python
book_genres = pd.read_csv("F:/Intro_DS/Project/book-rec-2/visualization/clean/book_genres.csv")
genres_name = pd.read_csv("F:/Intro_DS/Project/book-rec-2/visualization/clean/genres.csv")
```

```python
genre_count = defaultdict(int)
for genres in book_genres['genre_id']:
    genre_count[genres] += 1

# Sắp xếp tác giả theo count giảm dần
genre_count = sorted(genre_count.items(), key=lambda x: x[1], reverse=True)

# In kết quả
for genre, count in genre_count:
    print(f"{genre}: {count}")
    
```

```python
genreCount = [item[1] for item in genre_count]
genreName = [item[0] for item in genre_count]
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(genreName, genreCount, color='blue', marker='o')

# Thêm số count vào mỗi marker (cả genreName và count)
for i, count in enumerate(genreCount):
    plt.text(genreName[i], count, f'{count}', ha='center', va='bottom', color='black', fontsize=12)

# Thêm tiêu đề và nhãn trục
plt.title('Scatter Plot of Genres vs Genre Count')
plt.xlabel('Genres')
plt.ylabel('Genre Count')

# Hiển thị biểu đồ
plt.show()
```

```python


# Vẽ biểu đồ cột (bar chart)
# plt.bar(sorted_genreCount, sorted_genreName, color='red', width= 20)
plt.barh (genreName, genreCount, color='red', height= 50)

for i, count in enumerate(genreCount):
    plt.text(count + 10, i, str(count), va='center', color='black', fontsize=12)

# Thêm tiêu đề và nhãn cho các trục
plt.title('Genre Distribution')
plt.xlabel('Genres Count')
plt.ylabel('Genre Name')

# Hiển thị biểu đồ
plt.show()
```

```python
n, bins, patches = plt.hist(genreCount, bins=100, color='skyblue', edgecolor='black')

# Thêm số liệu vào mỗi thanh
# for i in range(len(patches)):
#     height = patches[i].get_height()  # Lấy chiều cao của thanh (tần suất)
#     plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, str(int(height)),
#              ha='center', va='bottom', color='black', fontsize=12)

# Thêm tiêu đề và nhãn cho các trục
plt.title('Histogram of Genre Counts')
plt.xlabel('Number of book with genres')
plt.ylabel('Count')

# Hiển thị biểu đồ
plt.show()
```

Có vài thể loại rất phổ thông, được viết nhiều, ngoài ra các thể loại khác khá đặc thù nên số lượng sách ít 

```python
genre_1 = 0
genre_2 = 0
genre_3 = 0
genre_4 = 0
genre_5 = 0
# i1 = 50
# i2 = 300
# i3 = 2500
# i4 = 4000

i1 = 50
i2 = 300
i3 = 1000
i4 = 4000


for genre, count in genre_count:
    if count <= i1:
        genre_1  += 1
    elif i1 < count <= i2:
        genre_2  += 1
    elif i2 < count <= i3:
        genre_3  += 1
    elif i3 < count <= i4:
        genre_4  += 1
    else:
        genre_5  += 1
        
print(f"{genre_1} {genre_2} {genre_3} {genre_5}")

# Danh sách các nhóm và số lượng tương ứng
labels = [f"0 - {i1}", f"{i1} - {i2}", f"{i2} - {i3}", f"{i3} - {i4}", f">{i4}"]
sizes = [genre_1, genre_2, genre_3,  genre_4, genre_5]

# Vẽ biểu đồ tròn
# plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.bar(labels, sizes, color = 'blue')
plt.xlabel("Number of book with genres")
plt.ylabel("Count")
# Tiêu đề biểu đồ
plt.title('Distribution of Genre Based on Book Count')
for i, size in enumerate(sizes):
    plt.text(i, size + 0.1, str(size), ha='center', va='bottom', fontsize=10)
# Hiển thị biểu đồ
plt.show()
```

##### 3. Review distribution

```python
reviews = pd.read_csv("F:/Intro_DS/Project/book-rec-2/visualization/clean/reviews.csv")
```

```python
review_count = reviews['rating_score'].value_counts()
review_count_df = review_count.reset_index()
review_count_df.columns = ['rating_score', 'count']

print(review_count_df)
```

```python
reviewCount = review_count.index  # Các giá trị rating_score
reviewName = review_count.values  # Số lần xuất hiện của từng giá trị

# Vẽ biểu đồ bar
plt.bar(review_count_df['rating_score'], review_count_df['count'], color='skyblue')
plt.xlabel('Rating Score')
plt.ylabel('Count')
plt.title('Distribution of Rating Scores')
plt.xticks(rotation=45)  # Xoay nhãn trục x nếu cần
plt.show()
```

Looking at the chart, users tend to give ratings of 3, 4, or 5 stars, while the other ratings are relatively low. It can be inferred that if a book is decent, users are likely to rate it, whereas if it's bad, they often ignore it.
