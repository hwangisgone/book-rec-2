```python
import pandas as pd
import json
import matplotlib.pyplot as plt
```

```python
data = pd.read_json('F:/Intro_DS/Project/book-rec-2/goodreadsscraper/goodreadsscraper/data/book/data_1_1000.json')
print(data)
```

```python
language_counts = data['language'].value_counts()
print(language_counts)
```

```python
# Tạo biểu đồ đường
plt.figure(figsize=(8, 5))
plt.plot(data['isbn'], data['ratingsCount'], marker='o', color='green', label='Ratings Count')

# Thêm tiêu đề và nhãn
plt.title('Ratings Count Trend', fontsize=16)
plt.xlabel('Book Title', fontsize=12)
plt.ylabel('Ratings Count', fontsize=12)
plt.legend()
plt.grid(True)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Hiển thị biểu đồ
plt.show()
```
