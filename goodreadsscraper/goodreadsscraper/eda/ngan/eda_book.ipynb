```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the JSON data from the file
with open('book.json', 'r') as file:
    data = pd.read_json(file)
```

```python
data
```

```python
# Tạo DataFrame
df = pd.DataFrame(data)

# Tách cột ratingHistogram thành 5 cột tương ứng
rating_columns = ['1 star', '2 star', '3 star', '4 star', '5 star']
df[rating_columns] = pd.DataFrame(df['ratingHistogram'].tolist(), index=df.index)

# Xóa cột ratingHistogram ban đầu (tuỳ chọn)
df.drop(columns=['ratingHistogram'], inplace=True)

df
```

```python
df = pd.DataFrame(df)

# Ensure isbn and isbn13 are string types
df['isbn'] = df['isbn'].astype(str)
df['isbn13'] = df['isbn13'].astype(str)

# Drop duplicates based on 'isbn' or 'isbn13'
df_unique = df.drop_duplicates(subset=['isbn', 'isbn13'])
df_unique
```

```python
unique_genres = set(
    genre for sublist in df['genres'].dropna().apply(lambda x: x if isinstance(x, list) else [])
    for genre in sublist
)

print("Unique genres:", unique_genres)
print("Number of unique genres:", len(unique_genres))
```

```python
# Flatten và thống kê genre từ DataFrame
all_genres = [genre for sublist in df['genres'].dropna().apply(lambda x: x if isinstance(x, list) else []) for genre in sublist]

# Thống kê số lượng sách theo thể loại
genre_counts = pd.Series(all_genres).value_counts()
top_genres = genre_counts.head(20)  # Lấy 20 thể loại phổ biến nhất
```

```python
# Vẽ biểu đồ
plt.figure(figsize=(12, 8))
plt.bar(top_genres.index, top_genres.values, color='teal')
plt.xticks(rotation=45)
plt.xlabel('Genres')
plt.ylabel('Number of Books')
plt.title('Top 20 Popular Genres')
plt.tight_layout()
plt.show()
```
