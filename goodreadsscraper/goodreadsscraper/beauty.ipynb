```python
import pandas as pd
import json
import matplotlib.pyplot as plt
from collections import Counter
```

```python
# Làm đẹp json 

# Đọc file JSON
with open('data/user/user_1_10.json', "r") as file:
    data = json.load(file)
# Ghi file JSON đẹp
with open("data/user/user_1_10_beauty.json", "w") as file:
    json.dump(data, file, indent=4, ensure_ascii=False)
```

```python


# Đường dẫn đến file JSON
file_path = 'F:/Intro_DS/Project/book-rec-2/goodreadsscraper/goodreadsscraper/data/review/review_1_10k.json'
rating_counts = []

# Mở file và xử lý từng dòng
with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

for index, book in enumerate(data, start=1):
    rating_count = len(book.get("rating", []))
    
    print(f"Sách {book.get('isbn')}: Số lượng ratings = {rating_count}")
    rating_counts.append(rating_count)
```

```python

# Đếm số lần xuất hiện của mỗi số
count = Counter(rating_counts)

# Tách các giá trị và tần suất ra thành hai danh sách
unique_numbers = list(count.keys())
frequencies = list(count.values())

# Vẽ biểu đồ cột
plt.bar(unique_numbers, frequencies)

# Thêm tiêu đề và nhãn cho trục
plt.title("Frequency of Each Number")
plt.xlabel("Number")
plt.ylabel("Frequency")

# Hiển thị biểu đồ
plt.show()
```

```python
isbns = []
for i in isbn_book:
    if i not in isbns:
        isbns.append(i)
        
print(len(isbns))
```

```python
file_input = "data/book/book_1_10k_clean_1.json"
crawl_sources = {}

# Mở file và xử lý từng dòng
with open(file_input, 'r', encoding='utf-8') as file:
    data = json.load(file)

# Giả sử isbns đã được định nghĩa
# isbns = ["isbn_1", "isbn_2", "isbn_3"]  # Thay thế bằng danh sách ISBN thực tế

for book in data:
    for isbn in isbns:
        if book.get("isbn") == isbn:
            source = book.get("crawlSource")
            if source not in crawl_sources:
                crawl_sources[source] = {"source": source}
                print(crawl_sources[source])

# In ra crawl_sources để xem kết quả
print(crawl_sources)
```
